```sql
SELECT COUNT(*) AS total_accounts
FROM bronze.take_home_test.accounts;
```

```sql
SELECT
    account_id,
    COUNT(*) AS qty
FROM bronze.take_home_test.accounts
GROUP BY account_id
HAVING COUNT(*) > 1;
```

```sql
SELECT *
FROM bronze.take_home_test.accounts
WHERE account_id IS NULL OR TRIM(account_id) = '';
```

```sql
SELECT *
FROM bronze.take_home_test.accounts
WHERE customer_id IS NULL OR TRIM(customer_id) = '';
```

```sql
SELECT DISTINCT status
FROM bronze.take_home_test.accounts;
```

```sql
SELECT customer_id, COUNT(*) AS qty
FROM bronze.take_home_test.accounts
GROUP BY customer_id
HAVING COUNT(*) > 1;
```

```sql
SELECT
    LENGTH(REGEXP_REPLACE(TRIM(account_number), '[^0-9]', '')) AS digits_account_number,
    COUNT(*) AS qty
FROM bronze.take_home_test.accounts
GROUP BY 1
ORDER BY 1;
```

```sql
SELECT
    LENGTH(REGEXP_REPLACE(TRIM(account_branch), '[^0-9]', '')) AS digits_branch,
    COUNT(*) AS qty
FROM bronze.take_home_test.accounts
GROUP BY 1
ORDER BY 1;
```

```sql
SELECT
    LENGTH(REGEXP_REPLACE(TRIM(account_check_digit), '[^0-9Xx]', '')) AS digits_check_digit,
    COUNT(*) AS qty
FROM bronze.take_home_test.accounts
GROUP BY 1
ORDER BY 1;
```
