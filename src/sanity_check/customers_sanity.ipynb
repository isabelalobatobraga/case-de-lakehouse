```sql
SELECT COUNT(*) AS total_customers
FROM bronze.take_home_test.customers;
```

```sql
SELECT
    customer_id,
    COUNT(*) AS qty
FROM bronze.take_home_test.customers
GROUP BY customer_id
HAVING COUNT(*) > 1;
```

```sql
SELECT
    REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '') AS cpf_normalized,
    COUNT(*) AS qty
FROM bronze.take_home_test.customers
GROUP BY cpf_normalized
HAVING COUNT(*) > 1;
```

```sql
WITH validator_cpf (
SELECT
    cpf,
    REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '') AS cpf_normalized,
    LENGTH(REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '')) AS digits,
    CASE 
        WHEN cpf IS NULL OR TRIM(cpf) = '' THEN 'EMPTY_OR_NULL'
        WHEN LENGTH(REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '')) < 11 THEN 'SHORT'
        WHEN LENGTH(REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '')) = 11 THEN 'VALID_LENGTH'
        WHEN LENGTH(REGEXP_REPLACE(TRIM(cpf), '[^0-9]', '')) > 11 THEN 'LONG'
        ELSE 'UNKNOWN'
    END AS status
FROM bronze.take_home_test.customers
)
SELECT 
  COUNT(*) AS qty_filter
FROM validator_cpf 
WHERE status = 'SHORT' OR status = 'VALID_LENGTH';
```
