```sql
SELECT
    id,
    COUNT(*) AS qty
FROM bronze.take_home_test.transfer_ins
GROUP BY id
HAVING COUNT(*) > 1;
```

```sql
SELECT *
FROM bronze.take_home_test.transfer_ins
LIMIT 20
```

```sql
SELECT DISTINCT status
FROM bronze.take_home_test.transfer_ins
```

```sql
SELECT *
FROM bronze.take_home_test.transfer_ins
WHERE status = "failed" and transaction_completed_at IS NOT NULL
```

```sql
SELECT
    id,
    COUNT(*) AS qty_rows,
    COUNT(DISTINCT transaction_completed_at) AS distinct_completed_values
FROM bronze.take_home_test.transfer_ins
GROUP BY id
HAVING COUNT(DISTINCT transaction_completed_at) > 1;
```
