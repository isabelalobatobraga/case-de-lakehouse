# Data Transformation – Gold Layer
## dim_city

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE TABLE IF NOT EXISTS gold.take_home_test.dim_city (
    city_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    city_id STRING,
    city_name STRING,
    inserted_at TIMESTAMP
);
```

```sql
MERGE INTO gold.take_home_test.dim_city AS dim
USING (
    SELECT DISTINCT
        city_id,
        city_name,
        current_timestamp() AS inserted_at
    FROM silver.take_home_test.city
) tbl
ON dim.city_id = tbl.city_id

WHEN NOT MATCHED THEN
INSERT (city_id, city_name, inserted_at)
VALUES (tbl.city_id, tbl.city_name, tbl.inserted_at);
```
