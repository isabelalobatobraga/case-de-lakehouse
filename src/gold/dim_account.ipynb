# Data Transformation – Gold Layer
## dim_account

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE TABLE IF NOT EXISTS gold.take_home_test.dim_account (
    account_sk        BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    account_id        STRING,
    account_code      STRING,
    account_code_type STRING,
    is_active         BOOLEAN,
    inserted_at       TIMESTAMP
);
```

```sql
MERGE INTO gold.take_home_test.dim_account AS dim
USING (
    SELECT DISTINCT
        account_id,
        account_code,
        account_code_type,
        is_active,
        current_timestamp() AS inserted_at
    FROM silver.take_home_test.accounts
) tbl
ON dim.account_id = tbl.account_id

WHEN NOT MATCHED THEN
INSERT (account_id, account_code, account_code_type, is_active, inserted_at)
VALUES (tbl.account_id, tbl.account_code, tbl.account_code_type, tbl.is_active, tbl.inserted_at);
```
