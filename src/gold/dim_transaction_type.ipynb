# Data Transformation – Gold Layer
## dim_transaction_type

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE OR REPLACE TABLE gold.take_home_test.dim_transaction_type (
    transaction_type_sk BIGINT GENERATED ALWAYS AS IDENTITY,
    transaction_method  STRING,
    movement_type       STRING,
    inserted_at         TIMESTAMP
);
```

```sql
INSERT INTO gold.take_home_test.dim_transaction_type (
    transaction_method, movement_type, inserted_at
)
SELECT DISTINCT
    transaction_method,
    movement_type,
    current_timestamp() AS inserted_at
FROM silver.take_home_test.transactions
ORDER BY transaction_method, movement_type;
```
