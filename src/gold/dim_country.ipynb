# Data Transformation – Gold Layer
## dim_country

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE TABLE IF NOT EXISTS gold.take_home_test.dim_country (
    country_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    country_id STRING,
    country_name STRING,
    inserted_at TIMESTAMP
);
```

```sql
MERGE INTO gold.take_home_test.dim_country AS dim
USING (
    SELECT DISTINCT
        country_id,
        country_name,
        current_timestamp() AS inserted_at
    FROM silver.take_home_test.country
) tbl
ON dim.country_id = tbl.country_id

WHEN NOT MATCHED THEN
INSERT (country_id, country_name, inserted_at)
VALUES (tbl.country_id, tbl.country_name, tbl.inserted_at);
```
