# Data Transformation – Gold Layer
## dim_customer

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE TABLE IF NOT EXISTS gold.take_home_test.dim_customer (
    customer_sk    BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    customer_id    STRING,
    first_name     STRING,
    last_name      STRING,
    document_type  STRING,
    document_number STRING,
    inserted_at    TIMESTAMP
);
```

```sql
MERGE INTO gold.take_home_test.dim_customer AS dim
USING (
    SELECT DISTINCT
        customer_id,
        first_name,
        last_name,
        "CPF" AS document_type,
        cpf_masked AS document_number,
        current_timestamp() AS inserted_at
    FROM silver.take_home_test.customers
) tbl
ON dim.customer_id = tbl.customer_id

WHEN NOT MATCHED THEN
INSERT (customer_id, first_name, last_name, document_type, document_number, inserted_at)
VALUES (tbl.customer_id, tbl.first_name, tbl.last_name, tbl.document_type, tbl.document_number, tbl.inserted_at);
```
