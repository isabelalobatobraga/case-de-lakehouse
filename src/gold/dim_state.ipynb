# Data Transformation – Gold Layer
## dim_state

This notebook performs the **dimensional modeling transformations** that populate the **gold layer**, generating **business-ready tables** such as dimensions and fact tables in **Delta Lake** format.

```sql
CREATE TABLE IF NOT EXISTS gold.take_home_test.dim_state (
    state_sk BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
    state_id STRING,
    state_code STRING,
    inserted_at TIMESTAMP
);
```

```sql
MERGE INTO gold.take_home_test.dim_state AS dim
USING (
    SELECT DISTINCT
        state_id,
        state_code,
        current_timestamp() AS inserted_at
    FROM silver.take_home_test.state
) tbl
ON dim.state_id = tbl.state_id

WHEN NOT MATCHED THEN
INSERT (state_id, state_code, inserted_at)
VALUES (tbl.state_id, tbl.state_code, tbl.inserted_at);
```
